# Python Operators

## Python Operators
Operators are used to perform operations on variables and values.
In the example below, we use the + operator to add together two values:

```python
# Example
print(10 + 5)
```

Although the `+` operator is often used to add together two values, like in the example above, it can also be used to add together a variable and a value, or two variables:

```python
# Example
sum1 = 100 + 50      # 150 (100 + 50)
sum2 = sum1 + 250    # 400 (150 + 250)
sum3 = sum2 + sum2   # 800 (400 + 400)
```

Python divides the operators in the following groups:

- Arithmetic operators
- Assignment operators
- Comparison operators
- Logical operators
- Identity operators
- Membership operators
- Bitwise operators

# Python Arithmetic Operators

## Arithmetic Operators
Arithmetic operators are used with numeric values to perform common mathematical operations:

|OPERATOR|NAME|EXAMPLE|
|---|---|---|
|+	|Addition|x + y	
|-	|Subtraction|x - y	
|*	|Multiplication|x * y	
|/	|Division|x / y	
|%	|Modulus|x % y	
|**|Exponentiation|x ** y	
|//|Floor division|x // y

```python
# Examples
# Here is an example using different arithmetic operators:


x = 15
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x ** y)
print(x // y)
```

## Division in Python
 Python has two division operators:

`/` - Division (returns a float) <br><br/>
`//` - Floor division (returns an integer)

```python
# Example
# Division always returns a float:

x = 12
y = 5

print(x / y)
```

```python
# Example
# Floor division always returns an integer.

# It rounds DOWN to the nearest integer:

x = 12
y = 5

print(x // y)
```

# Assignment Operators
>Assignment operators are used to assign values to variables:

| Operator | Example | Same As | Description |
| --- | --- | --- | --- |
| `=` | `x = 5` | `x = 5` | **Simple assignment:** assigns the value on the right to the variable on the left. |
| `+=` | `x += 3` | `x = x + 3` | **Addition assignment:** adds the value to the variable and assigns the result back. |
| `-=` | `x -= 3` | `x = x - 3` | **Subtraction assignment:** subtracts the value from the variable and assigns the result back. |
| `*=` | `x *= 3` | `x = x * 3` | **Multiplication assignment:** multiplies the variable by the value and assigns the result back. |
| `/=` | `x /= 3` | `x = x / 3` | **Division assignment:** divides the variable by the value (returns a float) and assigns the result back. |
| `%=` | `x %= 3` | `x = x % 3` | **Modulus assignment:** calculates the remainder of division and assigns it back to the variable. |
| `//=` | `x //= 3` | `x = x // 3` | **Floor division assignment:** divides, truncates the decimal part, and assigns the integer result back. |
| `**=` | `x **= 3` | `x = x ** 3` | **Exponentiation assignment:** raises the variable to the specified power and assigns the result back. |
| `&=` | `x &= 3` | `x = x & 3` | **Bitwise AND assignment:** performs a bitwise `AND` operation and assigns the result back. |
| `|=` | `x |= 3` | `x = x | 3` | **Bitwise OR assignment:** performs a bitwise `OR` operation and assigns the result back. |
| `^=` | `x ^= 3` | `x = x ^ 3` | **Bitwise XOR assignment:** performs a bitwise `XOR` (exclusive OR) operation and assigns the result back. |
| `>>=` | `x >>= 3` | `x = x >> 3` | **Right shift assignment:** shifts the bits to the right by the specified amount and assigns the result back. |
| `<<=` | `x <<= 3` | `x = x << 3` | **Left shift assignment:** shifts the bits to the left by the specified amount and assigns the result back. |
| `:=` | `print(x := 3)` | `x = 3; print(x)` | **Walrus operator (Assignment expression):** assigns a value to a variable as part of a larger expression while simultaneously returning that value. |

## The Walrus Operator
Python 3.8 introduced the `:=` operator, known as the "walrus operator". It assigns values to variables as part of a larger expression:

```python
# Example
# The count variable is assigned in the if statement, and given the value 5:

numbers = [1, 2, 3, 4, 5]

if (count := len(numbers)) > 3:
    print(f"List has {count} elements")
```

# Python Ternary Operator

## The Ternary Operator
The ternary operator allows you to assign one value if a condition is true, and another if it is false:



```python
# Example
# Assign a value to x:
# Assign the value "WEEKEND!" if the number is higher than 5, otherwise "Workday":

num = 6
x = "WEEKEND!" if num > 5 else "Workday"
print(x)
```

**Note:** The ternary operator is not an actual operator, it is a conditional expression, or a shorthand if statement.

## Instead of Elif:
The ternary operator can be used instead of elif in longer if statements:

```python
# Example
# Assign:
# - "Fri" if num is 5
# - "Sat" if num is 6
# - "Sun" if num is 7
# - otherwise assign "weekday":

num = 6
x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "weekday"
print(x)
```

# Python Comparison Operators

## Comparison Operators
Comparison operators are used to compare two values:

|Operator|Name|Example|
|---|---|---|
|==|Equal|x == y|
|!=|Not equal|x != y|
|>|Greater than|x > y|
|<|Less than|x < y|
|>=|Greater than or equal to|x >= y	|

```python
# Examples
# Comparison operators return True or False based on the comparison:

x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Example

x = 5
print(1 < x < 10)
print(1 < x and x < 10)
```

# Python Logical Operators

## Logical Operators
Logical operators are used to combine conditional statements:

|Operator|Description|Example|
|---|---|---|
|and|Returns True if both statements are true|x < 5 and  x < 10|
|or|Returns True if one of the statements is true|x < 5 or x < 4|
|not|Reverse the result, returns False if the result is true|not(x < 5 and x < 10)|


```python
# Example
# Test if a number is greater than 0 and less than 10:

x = 5
print(x > 0 and x < 10)
```

```python
# Example
# Test if a number is less than 5 or greater than 10:

x = 5
print(x < 5 or x > 10)
```

```python
# Example
# Reverse the result with not:

x = 5
print(not(x > 3 and x < 10))
```

# Python Identity Operators

## Identity Operators
> Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:

|Operator|Description|Example|
|---|---|---|
|is|Returns True if both variables are the same object|x is y|	
|is not|Returns True if both variables are not the same object|x is not y|

```python
# Example
# The is operator returns True if both variables point to the same object:

x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
```

```python
# Example
# The is not operator returns True if both variables do not point to the same object:

x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is not y)
```

## Difference Between is and ==
- **is** - Checks if both variables point to the same object in memory
- `==` - Checks if the values of both variables are equal

# Python Membership Operators

## Membership Operators
Membership operators are used to test if a sequence is presented in an object:

|Operator|Description|Example|
|---|---|---|
|in|Returns True if a sequence with the specified value is present in the object|x in y|
|not in|Returns True if a sequence with the specified value is not present in the object|x not in y|

```python
# Example
# Check if "banana" is present in a list:

fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
```

```python
# Example
# Check if "pineapple" is NOT present in a list:

fruits = ["apple", "banana", "cherry"]
print("pineapple" not in fruits)
```

## Membership in Strings
> The membership operators also work with strings:

```python
# Example

text = "Hello World"

print("H" in text)
print("hello" in text)
print("z" not in text)
```

# Python Bitwise Operators

## Bitwise Operators
Bitwise operators are used to compare (binary) numbers:

| Operator | Name | Description | Example |
| :---: | :--- | :--- | :--- |
| `&` | **AND** | Sets each bit to `1` if both bits are `1`. | `5 & 3` $\rightarrow$ `1`<br>`(0101 & 0011 = 0001)` |
| `\|` | **OR** | Sets each bit to `1` if at least one of the bits is `1`. | `5 \| 3` $\rightarrow$ `7`<br>`(0101 \| 0011 = 0111)` |
| `^` | **XOR** | Sets each bit to `1` only if one bit is `1` and the other is `0` (exclusive OR). | `5 ^ 3` $\rightarrow$ `6`<br>`(0101 ^ 0011 = 0110)` |
| `~` | **NOT** | Inverts all the bits (bitwise negation / ones' complement). In Python, `~x` evaluates to `-(x + 1)`. | `~5` $\rightarrow$ `-6`<br>`(~0101 = ...1010)` |
| `<<` | **Zero-fill Left Shift** | Shifts bits to the left by the specified number of positions, filling empty spaces on the right with `0`. Effectively multiplies by $2^n$. | `5 << 1` $\rightarrow$ `10`<br>`(0101 << 1 = 1010)` |
| `>>` | **Signed Right Shift** | Shifts bits to the right by the specified number of positions, discarding bits that fall off. Effectively integer-divides by $2^n$. | `5 >> 1` $\rightarrow$ `2`<br>`(0101 >> 1 = 0010)` |

```python
# Example
# The & operator compares each bit and set it to 1 if both are 1, otherwise it is set to 0:

print(6 & 3)
# The binary representation of 6 is 0110
# The binary representation of 3 is 0011
# Then the & operator compares the bits and returns 0010, which is 2 in decimal.
```

```python
# Example
# The | operator compares each bit and set it to 1 if one or both is 1, otherwise it is set to 0:

print(6 | 3)
# The binary representation of 6 is 0110
# The binary representation of 3 is 0011

# Then the | operator compares the bits and returns 0111, which is 7 in decimal.
```

```python
# Example
# The ^ operator compares each bit and set it to 1 if only one is 1, otherwise (if both are 1 or both are 0) it is set to 0:

print(6 ^ 3)
# The binary representation of 6 is 0110
# The binary representation of 3 is 0011

# Then the ^ operator compares the bits and returns 0101, which is 5 in decimal.
```

# Operator Precedence

Operator precedence dictates the order in which Python evaluates different operations within an expression. When an expression contains multiple operators, precedence rules act as the grammar that prevents ambiguity.

* **Precedence (Priority Level):** Operators with higher precedence are evaluated before operators with lower precedence. For instance, multiplication (`*`) has higher precedence than addition (`+`), meaning $2 + 3 \times 4$ evaluates to $14$, not $20$.
* **Associativity (Direction of Evaluation):** When an expression contains two or more operators of the **same** precedence level, associativity determines whether they are resolved from left to right or right to left:
  * **Left-to-Right Associativity:** Most Python operators (such as addition, subtraction, division, and multiplication) evaluate from left to right: `10 - 4 - 2` evaluates as `(10 - 4) - 2 = 4`.
  * **Right-to-Left Associativity:** The exponentiation operator (`**`) and conditional expressions evaluate from right to left: `2 ** 3 ** 2` evaluates as `2 ** (3 ** 2) = 2 ** 9 = 512`.
* **Explicit Grouping with Parentheses `()`:** Parentheses have the highest precedence of all syntactic constructs. Surrounding an expression with `()` forces the enclosed sub-expression to be evaluated first, overriding standard operator ranking.
* **Chained Comparisons:** Python evaluates chained comparison operators (such as `a < b <= c`) as `(a < b) and (b <= c)`, evaluating each term with implicit short-circuit logic while maintaining identical comparison priority.
* **Logical Operator Hierarchy:** Within boolean logic, operators do not hold equal rank: `not` evaluates before `and`, and `and` evaluates before `or`.
