# 1. Python Strings
---
## Strings
Strings in python are surrounded by either single quotation marks, or double quotation marks.
- 'hello' is the same as "hello".
You can display a string literal with the print() function:

```python
# Example
print("Olá")
print('Hola')
print('Hello')
```

## Quotes Inside Quotes
You can use quotes inside a string, as long as they don't match the quotes surrounding the string:

```python
# Example
print("It's alright hombre")
print("He is called 'Juan'")
print('He is called "Juanito"')
```

## Assign String to a Variable
> Assigning a string to a variable is done with the variable name followed by an equal sign and the string:

```python
a = "Hello"
print(a)
```

## Multiline Strings
You can assign a multiline string to a variable by using three quotes:

```python
# Example
# You can use three double quotes:

a = """Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)
```

```python
# Or three single quotes:

a = '''Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua.'''
print(a)
```

Note: in the result, the line breaks are inserted at the same position as in the code.

## Strings are Arrays
Like many other popular programming languages, strings in Python are arrays of unicode characters.
However, Python does not have a character data type, a single character is simply a string with a length of 1.
Square brackets can be used to access elements of the string.

```python
# Example
# Get the character at position 1 (remember that the first character has the position 0):

a = "Hello, World!"
print(a[1])
```

## Looping Through a String
Since strings are arrays, we can loop through the characters in a string, with a `for` loop.

```python
# Example
# Loop through the letters in the word "banana":

for x in "banana":
  print(x)
```

## String Length
To get the length of a string, use the `len()` function.

```python
# Example
# The len() function returns the length of a string:

a = "Vamos virar, Python!"
print(len(a))
```

## Check String
To check if a certain phrase or character is present in a string, we can use the keyword in.

```python
# Example
# Check if "free" is present in the following text:

txt = "Here you come again, restless shadows. - Dom Casmurro, Machado de Assis."
print("Assis" in txt)
```

```python
# Use it in an if statement:


# Example
# Print only if "free" is present:

txt = "Here you come again, restless shadows. - Dom Casmurro, Machado de Assis."
x = "shadows"
if x in txt:
  print(f"Yes, {x} is present.")
```

## Check if NOT
To check if a certain phrase or character is NOT present in a string, we can use the keyword `not in`.

```python
# Example
# Check if "expensive" is NOT present in the following text:

txt = "Here you come again, restless shadows. - Dom Casmurro, Machado de Assis."
x = "shadows"
print(x not in txt)
```

```python
# Use it in an if statement:

# Example
# print only if "expensive" is NOT present:

txt = "Here you come again, restless shadows. - Dom Casmurro, Machado de Assis."
x = "soul"
if x not in txt:
  print(f"No, '{x}' is NOT present.")
```

# 2. Python - Slicing Strings
---
## Slicing
You can return a range of characters by using the slice syntax.
Specify the start index and the end index, separated by a colon, to return a part of the string.

**Note:** The first character has index 0.

```python
# Example
# Get the characters from position 2 to position 5 (not included):

b = "Olá, mundo!"
print(b[2:5])
```

## Slice From the Start
By leaving out the start index, the range will start at the first character:

```python
# Example
# Get the characters from the start to position 5 (not included):

b = "Olá, mundo!"
print(b[:5])
```

## Slice To the End
By leaving out the end index, the range will go to the end:

```python
# Example
# Get the characters from position 2, and all the way to the end:

b = "Olá, mundo!"
print(b[2:])
```

## Negative Indexing
Use negative indexes to start the slice from the end of the string:

```python
# Example
# Get the characters:
# From: "o" in "World!" (position -5)
# To, but not included: "d" in "World!" (position -2):

b = "Hello, World!"
print(b[-5:-2])
```

# 3. Python - Modify Strings
Python has a set of built-in methods that you can use on strings.


## Upper Case

```python
# Example
# The upper() method returns the string in upper case:

a = "Hello, World!"
print(a.upper())
```

## Lower Case

```python
# Example
# The lower() method returns the string in lower case:

a = "Hello, World!"
print(a.lower())
```

## Remove Whitespace
Whitespace is the space before and/or after the actual text, and very often you want to remove this space.

```python
# Example
# The strip() method removes any whitespace from the beginning or the end:

a = " Hello, World! "
print(a.strip()) # returns "Hello, World!"
```

## Replace String

```python
# Example
# The replace() method replaces a string with another string:

a = "Hello, World!"
print(a.replace("H", "J"))
```

## Split String
The `split()` method returns a list where the text between the specified separator becomes the list items.

```python
# Example
# The split() method splits the string into substrings if it finds instances of the separator:

a = "Hello, World!"
print(a.split(",")) # returns ['Hello', ' World!']
```

# 4. Python - String Concatenation
---
## String Concatenation
To concatenate, or combine, two strings you can use the + operator.

```python
# Example
# Merge variable a with variable b into variable c:

a = "Hello"
b = "World"
c = a + b
print(c)
```

```python
# Example
# To add a space between them, add a " ":

a = "Hello"
b = "World"
c = a + " " + b
print(c)
```
