# Python Variables

## Variables
>Variables are containers for storing data values.

## Creating Variables
Python has no command for declaring a variable.
A variable is created the moment you first assign a value to it.

```python
# Example
x = 9
y = "Jay"
print(x)
print(y)
```

Variables do not need to be declared with any particular type, and can even change type after they have been set.

```python
#Example
x = 9       # x is of type int
x = "Katherine" # x is now of type str
print(x)
```

## Casting
If you want to specify the data type of a variable, this can be done with casting.

```python
x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
```

## Get the Type

>You can get the data type of a variable with the `type()` function.



```python
x = 9
y = "Katherine"
print(type(x))
print(type(y))
```

- Variables can be declared either by using single or double quotes.
- Varibles are case-sensitive: `A is different of a.`

# Python - Variable Names

## Variable Names
>A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume).

## Rules for Python Variables
- A variable name must start with a letter or the underscore character
- A variable name cannot start with a number
- A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
- Variable names are case-sensitive (age, Age and AGE are three different variables)
- A variable name cannot be any of the Python keywords.

```python
# Example
# Legal variable names:

myvar = "Mila"
my_var = "Mila"
_my_var = "Mila"
myVar = "Mila"
MYVAR = "Mila"
myvar2 = "Mila"

# Illegal variable names:
# 2myvar = "Mila"
# my-var = "Mila"
# my var = "Mila"
```

## Multi Words Variable Names
Variable names with more than one word can be difficult to read.
There are several techniques you can use to make them more readable:

### Camel Case
Each word, except the first, starts with a capital letter:
`myVariableName = "John"`

### Pascal Case
Each word starts with a capital letter:
`MyVariableName = "John"`

### Snake Case
Each word is separated by an underscore character:
`my_variable_name = "John"`

## Python Variables - Assign Multiple Values

### Many Values to Multiple Variables
Python allows you to assign values to multiple variables in one line:

```python
# Example
x, y, z = "Brasil", "Panamá", "México"
print(x)
print(y)
print(z)
```

**Note**: Make sure the number of variables matches the number of values, or else you will get an error.

### One Value to Multiple Variables
And you can assign the same value to multiple variables in one line:

```python
#Example

x = y = z = "Jabuticaba"
print(x)
print(y)
print(z)
```

### Unpack a Collection
If you have a collection of values in a list, tuple etc. Python allows you to extract the values into variables. This is called unpacking.

```python
colors = ["red", "yellow", "purple"]
x, y, z = colors
print(x)
print(y)
print(z)
```

# Python - Output Variables

Output Variables
The `print()` function is often used to output variables.

```python
x = "Python"
print(x)
```

In the `print()` function, you output multiple variables, separated by a comma:

```python
x = "Python"
y = "is"
z = "efficient"
print(x, y, z)

# Notice the space character after "Python " and "is ", 
# without them the result would be "Pythonisefficient".
```

For numbers, the + character works as a mathematical operator:

```python
x = 5
y = 10
print(x + y)
```

In the `print()` function, when you try to combine a string and a number with the + operator, Python will give you an error:

```python
#Example
x = 5
y = "John"
print(x + y)
```

The best way to output multiple variables in the `print()` function is to separate them with commas, which even support different data types:

```python
x = 5
y = "John"
print(x, y)
```

# Python - Global Variables

## Global Variables
> Variables that are created outside of a function (as in all of the examples in the previous pages) are known as global variables.

Global variables can be used by everyone, both inside of functions and outside.

```python
#Example
#Create a variable outside of a function, and use it inside the function

# Global variable - can be used inside of functions and outside
x = "efficient"

# Python Function
def myfunc():
  print("Python is " + x)
# Calling Python function
myfunc()
```

- If you create a variable with the same name inside a function, this variable will be local, and can only be used inside the function. 

- The global variable with the same name will remain as it was, global and with the original value.

```python
# Example
# Create a variable inside a function, with the same name as the global variable

# global variable
x = "awesome"

# function block starts here
def myfunc():
  # local variable
  x = "fantastic"
  print("Python is " + x) # function block ends here

# local variable output
myfunc()

# global variable output
print("Python is " + x)
```

## The global Keyword

>Normally, when you create a variable inside a function, that variable is local, and can only be used inside that function.

To create a global variable inside a function, you can use the `global` keyword.

```python
# Example
# If you use the global keyword, the variable belongs to the global scope:

def myfunc():
  # global variable
  global x
  x = "fantastic"

myfunc()

print("Python is " + x)
```

Also, use the global keyword if you want to change a global variable inside a function.

```python
# Example
# To change the value of a global variable inside a function, refer to the variable by using the global keyword:

x = "awesome"

def myfunc():
  global x
  # assingning a new value to a global variable - awesome to fantastic
  x = "fantastic"

myfunc()

print("Python is " + x)
```
